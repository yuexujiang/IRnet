```python
# alternative to block 2 and 3
keg2sym={}
sym2keg={}
f = open("./whole_Entrez.txt", "r")
for x in f:
    sym=x.strip().split()[0]
    kegid="hsa:"+x.strip().split()[1]
    keg2sym[kegid]=sym
    sym2keg[sym]=kegid
    
col=[]
col_sym=[]
with open('./genelist_8080.txt','r') as reader:
    line = reader.readline().strip()
    while line != '':  # The EOF char is an empty string
        col_sym.append(line)
        col.append(sym2keg[line])
        line = reader.readline().strip()
    
```

```python
# tide_data/Braun

def loadData(x,y,info,noteString,file_exp,file_clinic):
    print(noteString)
    pt2info={}
    pt2res={}
    pt2exp={}
    f=open(file_clinic)
    line = f.readline()
    sep=line.strip().split("\t")
    ind=sep.index("response_NR")
    lines = f.readlines()
    l=np.max([len(lines[0].split("\t")),len(lines[1].split("\t")),len(lines[2].split("\t"))])
    print(l)
    if len(sep)<l:
        response_colnum=ind+1
    else:
        response_colnum=ind
    print("response_column: "+str(response_colnum))
    num=0
    for line in lines:
        sep=line.strip().split("\t")
        if response_colnum+1>len(sep):
            continue
        #print(sep)
        num+=1
        pt=sep[0]
        inf=noteString+"_"+pt
        res=float(sep[response_colnum])
        pt2info[pt]=inf
        pt2res[pt]=res
    print("sample num: "+str(num))
    f=open(file_exp)
    exp_title = f.readline().strip().split("\t")
    lines=f.readlines()
    exp=[]
    #l=len(lines[0].strip().split("\t"))
    for line in lines:
        line=line.split("\t")
        for i in range(len(line)):
            line[i]=line[i].strip()
            if line[i]=="":
                line[i]="0.0"
        if "NA" in line[2:]:   ########################
            continue
        else:
            exp.append(line)
    exp=np.array(exp)
    genes=exp[:,1]   ###########################
    exp=exp[:,2:]    #############################
    exp_order=np.zeros((len(col),int(exp.shape[1])))
    print("gene num: "+str(len(genes)))
    untouch_list = list(range(len(col_sym)))
    for i in range(len(genes)):
        if not genes[i] in col_sym:
            continue
        j=col_sym.index(genes[i])
        untouch_list.remove(j)
        exp_order[j,:]=exp[i,:]
    print(len(untouch_list))
    for pat_j in range(exp_order.shape[1]):
        v=[]
        for g_i in range(exp_order.shape[0]):
            if not g_i in untouch_list:
                v.append(exp_order[g_i, pat_j])
        mean_v = np.median(v)
        for g_i in untouch_list:
            exp_order[g_i, pat_j]=mean_v
    if len(exp_title)>exp_order.shape[1]:
        exp_title=exp_title[1:]
    for i in range(len(exp_title)):
        pid=exp_title[i]    ######################
        pt2exp[pid]=exp_order[:,i]
    for k in pt2exp.keys():
        if k in pt2res.keys():
            x.append(pt2exp[k])
            y.append(pt2res[k])
            info.append(pt2info[k])
    print("\n")
    
x_test=[]
y_test=[]
info_test=[]

# file_exp="./tide_data/GSE176307/exp.txt"
# file_clinic="./tide_data/GSE176307/clinic.txt"
# noteString="GSE176307_BLCA_RNAseq"

# file_exp="./tide_data/Mariathasan/exp.txt"
# file_clinic="./tide_data/Mariathasan/clinic.txt"
# noteString="Mariathasan_BLCA_RNAseq"

# file_exp="./tide_data/Checkmate025/exp.txt"
# file_clinic="./tide_data/Checkmate025/clinic_Niv.txt"
# noteString="Checkmate025_KIRC_RNAseq"

file_exp="./tide_data/Braun/exp.tsv"
file_clinic="./tide_data/Braun/clinic.txt"
noteString="Braun_RCC_RNAseq"

response_colnum=7
loadData(x_test,y_test,info_test,noteString,file_exp,file_clinic)
```
